## Assignment 3: Data Cleaning

### Task 1

```python
df = pd.DataFrame(np.random.randint(1, 10, (5, 3)), columns=['A', 'B', 'C']).astype(float)
df.iloc[0, 1] = np.nan
df.iloc[2, 0] = np.nan
df.iloc[4, 2] = np.nan
print(df)
print(df.fillna(df.mean()))
```

### Task 2

```python
df = pd.DataFrame(np.random.randint(1, 10, (6, 4)), columns=list('ABCD')).astype(float)
df.iloc[1, 2] = np.nan
df.iloc[3, 0] = np.nan
print(df)
print(df.dropna())
```

## Assignment 4: Data Aggregation

### Task 1

```python
df = pd.DataFrame({'Category': np.random.choice(['A', 'B', 'C'], 10), 'Value': np.random.randint(1, 100, 10)})
print(df.groupby('Category')['Value'].agg(['sum', 'mean']))
```

### Task 2

```python
df = pd.DataFrame({'Product': np.random.choice(['P1', 'P2', 'P3', 'P4'], 10), 'Category': np.random.choice(['A', 'B', 'C'], 10), 'Sales': np.random.randint(100, 1000, 10)})
print(df.groupby('Category')['Sales'].sum())
```

## Assignment 5: Merging DataFrames

### Task 1

```python
df1 = pd.DataFrame({'ID': [1, 2, 3, 4], 'Name': ['Ali', 'Sara', 'Omar', 'Zoya']})
df2 = pd.DataFrame({'ID': [2, 3, 4, 5], 'Score': [88, 92, 75, 60]})
print(pd.merge(df1, df2, on='ID'))
```

### Task 2

```python
df1 = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
df2 = pd.DataFrame({'C': [7, 8, 9], 'D': [10, 11, 12]})
print(pd.concat([df1, df2], axis=0))
print(pd.concat([df1, df2], axis=1))
```

## Assignment 6: Time Series Analysis

### Task 1

```python
df = pd.DataFrame({'Value': np.random.randint(1, 100, 90)}, index=pd.date_range('2021-01-01', periods=90, freq='D'))
print(df.resample('MS').mean())
```

### Task 2

```python
df = pd.DataFrame({'Value': np.random.randint(1, 100, 365)}, index=pd.date_range('2021-01-01', '2021-12-31'))
print(df.rolling(window=7).mean().head(10))
```

## Assignment 7: MultiIndex DataFrame

### Task 1

```python
idx = pd.MultiIndex.from_product([['A', 'B'], [1, 2, 3]], names=['Letter', 'Number'])
df = pd.DataFrame({'Value': np.random.randint(1, 100, 6)}, index=idx)
print(df)
print(df.loc['A'])
print(df.loc[('B', 2)])
print(df.xs(1, level='Number'))
```

### Task 2

```python
idx = pd.MultiIndex.from_product([['Electronics', 'Clothing'], ['Sub1', 'Sub2']], names=['Category', 'SubCategory'])
df = pd.DataFrame({'Value': np.random.randint(1, 100, 4)}, index=idx)
print(df)
print(df.groupby(level=['Category', 'SubCategory']).sum())
```

## Assignment 8: Pivot Tables

### Task 1

```python
df = pd.DataFrame({'Date': np.repeat(pd.date_range('2021-01-01', periods=3), 2), 'Category': ['A', 'B'] * 3, 'Value': np.random.randint(1, 100, 6)})
print(df.pivot_table(values='Value', index='Date', columns='Category', aggfunc='sum'))
```

### Task 2

```python
df = pd.DataFrame({'Year': np.repeat([2021, 2022], 4), 'Quarter': ['Q1', 'Q2', 'Q3', 'Q4'] * 2, 'Revenue': np.random.randint(1000, 5000, 8)})
print(df.pivot_table(values='Revenue', index='Year', columns='Quarter', aggfunc='mean'))
```
