### Assignment 2: Array Indexing and Slicing

**Q1.** (6, 6) array (1–36), 3rd–5th rows aur 2nd–4th columns ka sub-array.

```python
arr = np.arange(1, 37).reshape(6, 6)
print("Array:\n", arr)

sub = arr[2:5, 1:4]   # rows 3-5 => index 2:5, cols 2-4 => index 1:4
print("\nSub-array (rows 3-5, cols 2-4):\n", sub)
```

**Q2.** (5, 5) random array, border ke elements extract karo.

```python
arr = np.random.randint(1, 100, size=(5, 5))
print("Array:\n", arr)

# Method 1: border elements ek 1D array me
border = np.concatenate([arr[0, :], arr[-1, :], arr[1:-1, 0], arr[1:-1, -1]])
print("\nBorder elements (1D):", border)

# Method 2: boolean mask se (interior ko hata kar)
mask = np.ones(arr.shape, dtype=bool)
mask[1:-1, 1:-1] = False
print("Border via mask      :", arr[mask])
print("\nBorder only (interior = 0):\n", np.where(mask, arr, 0))
```

**Q2.** (4, 4) array me se shape (4,) ka 1D array har column se subtract karo.

> Note: "har column se subtract" ka matlab hai ke 1D array ko column vector bana kar (`[:, np.newaxis]`) subtract kiya jaye. Row-wise version bhi neeche dikhaya hai.

```python
arr = np.random.randint(10, 50, size=(4, 4))
vec = np.array([1, 2, 3, 4])

print("2D array:\n", arr)
print("\n1D array:", vec)

# Har column me se vec subtract (vec ko column vector banao: shape (4,1))
print("\nSubtract from each column:\n", arr - vec[:, np.newaxis])

# Alternative: har row me se subtract (default broadcasting)
print("\n(Alternative) Subtract from each row:\n", arr - vec)
```

### Assignment 6: Linear Algebra

**Q1.** (3, 3) matrix ka determinant, inverse, eigenvalues.

```python
M = np.array([[2, 1, 1],
              [1, 3, 2],
              [1, 0, 0]])
print("Matrix:\n", M)

det = np.linalg.det(M)
print("\nDeterminant:", round(det, 4))

if not np.isclose(det, 0):
    inv = np.linalg.inv(M)
    print("\nInverse:\n", np.round(inv, 4))
    print("\nVerify M @ inv = I:\n", np.round(M @ inv, 4))
else:
    print("Matrix singular hai, inverse exist nahi karta.")

eigvals = np.linalg.eigvals(M)
print("\nEigenvalues:", np.round(eigvals, 4))
```

### Assignment 9: Structured Arrays

**Q1.** Fields `name`, `age`, `weight` ke saath structured array banao aur age ke hisaab se sort karo.

```python
dtype = [('name', 'U10'), ('age', 'i4'), ('weight', 'f4')]
people = np.array([('Ali',   25, 70.5),
                   ('Sara',  22, 55.2),
                   ('Ahmed', 30, 82.0),
                   ('Maryam',21, 58.3),
                   ('Usman', 27, 75.1)], dtype=dtype)

print("Original:\n", people)

sorted_people = np.sort(people, order='age')
print("\nSorted by age:\n", sorted_people)
```

**Q2.** Fields `x`, `y` (integers); har pair of points ke beech Euclidean distance.

```python
pts = np.array([(0, 0), (3, 4), (6, 8), (1, 1)], dtype=[('x', 'i4'), ('y', 'i4')])
print("Points:\n", pts)

x, y = pts['x'].astype(float), pts['y'].astype(float)

# Broadcasting se pairwise distance matrix
dx = x[:, None] - x[None, :]
dy = y[:, None] - y[None, :]
dist = np.sqrt(dx**2 + dy**2)
print("\nPairwise Euclidean distance matrix:\n", np.round(dist, 3))

# Pairs ki list
print()
n = len(pts)
for i in range(n):
    for j in range(i + 1, n):
        print(f"Point {i} ({x[i]:.0f},{y[i]:.0f}) <-> Point {j} ({x[j]:.0f},{y[j]:.0f}) : {dist[i, j]:.3f}")
```

**Q2.** (3, 3) masked array, diagonal mask karo, masked elements ko unmasked ke mean se replace karo.

```python
data = np.random.randint(1, 20, size=(3, 3))
mask = np.eye(3, dtype=bool)                 # diagonal True
masked = np.ma.array(data, mask=mask)

print("Original:\n", data)
print("\nMasked (diagonal hidden):\n", masked)

mean_val = masked.mean()
print("\nMean of unmasked elements:", round(float(mean_val), 4))

replaced = masked.filled(mean_val)
print("\nMasked elements replaced with mean:\n", np.round(replaced, 2))
```
