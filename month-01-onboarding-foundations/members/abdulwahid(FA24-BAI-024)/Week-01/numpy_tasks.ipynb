## Assignment 4: Statistical Operations

### Task 1

```python
a = np.random.randint(1, 50, (5, 5))
print(a.mean(), np.median(a), a.std(), a.var())
```

## Assignment 9: Structured Arrays

### Task 1

```python
dt = np.dtype([('name', 'U10'), ('age', 'i4'), ('weight', 'f4')])
people = np.array([('Ali', 25, 70.5), ('Sara', 22, 55.0), ('Omar', 30, 82.3)], dtype=dt)
print(np.sort(people, order='age'))
```

### Task 2

```python
pts = np.array([(0, 0), (3, 4), (6, 8)], dtype=[('x', 'i4'), ('y', 'i4')])
dx = pts['x'][:, None] - pts['x'][None, :]
dy = pts['y'][:, None] - pts['y'][None, :]
print(np.sqrt(dx**2 + dy**2))
```

## Assignment 10: Masked Arrays

### Task 1

```python
a = np.random.randint(1, 20, (4, 4))
m = np.ma.masked_greater(a, 10)
print(m)
print(m.sum())
```

### Task 2

```python
a = np.random.randint(1, 20, (3, 3)).astype(float)
m = np.ma.masked_array(a, mask=np.eye(3, dtype=bool))
print(m.filled(m.mean()))
```
