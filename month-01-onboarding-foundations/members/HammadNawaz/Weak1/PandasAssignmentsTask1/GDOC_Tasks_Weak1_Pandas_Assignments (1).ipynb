# Assignment 3: Data Cleaning

### Task 1: Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.

```python
np.random.seed(42)
df_clean1 = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)), columns=['A', 'B', 'C']).astype(float)
df_clean1.iloc[1, 0] = np.nan
df_clean1.iloc[3, 2] = np.nan
print("DataFrame with NaNs:")
display(df_clean1)
print("\nDataFrame filled with Column Mean:")
df_clean1_filled = df_clean1.fillna(df_clean1.mean())
display(df_clean1_filled)
```

### Task 2: Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Introduce some NaN values. Drop the rows with any NaN values.

```python
np.random.seed(42)
df_clean2 = pd.DataFrame(np.random.randint(1, 100, size=(6, 4)), columns=['A', 'B', 'C', 'D']).astype(float)
df_clean2.iloc[2, 1] = np.nan
df_clean2.iloc[4, 3] = np.nan
print("DataFrame with NaNs:")
display(df_clean2)
print("\nDataFrame after dropping rows with NaNs:")
df_clean2_dropped = df_clean2.dropna()
display(df_clean2_dropped)
```

# Assignment 4: Data Aggregation

### Task 1: Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.

```python
np.random.seed(42)
categories = np.random.choice(['A', 'B', 'C'], size=10)
values = np.random.randint(10, 100, size=10)
df_agg1 = pd.DataFrame({'Category': categories, 'Value': values})
print("Original DataFrame:")
display(df_agg1)
print("\nAggregated Sum and Mean:")
grouped1 = df_agg1.groupby('Category')['Value'].agg(['sum', 'mean'])
display(grouped1)
```

### Task 2: Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.

```python
np.random.seed(42)
products = [f'Prod_{i}' for i in range(1, 9)]
categories = np.random.choice(['Electronics', 'Clothing', 'Home'], size=8)
sales = np.random.randint(100, 1000, size=8)
df_agg2 = pd.DataFrame({'Product': products, 'Category': categories, 'Sales': sales})
print("Original DataFrame:")
display(df_agg2)
print("\nTotal Sales per Category:")
total_sales = df_agg2.groupby('Category')['Sales'].sum().reset_index()
display(total_sales)
```

# Assignment 5: Merging DataFrames

### Task 1: Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.

```python
df_left = pd.DataFrame({'Emp_ID': [101, 102, 103, 104], 'Name': ['Alice', 'Bob', 'Charlie', 'David']})
df_right = pd.DataFrame({'Emp_ID': [102, 103, 104, 105], 'Department': ['HR', 'IT', 'Finance', 'Marketing']})
print("Left DataFrame:")
display(df_left)
print("\nRight DataFrame:")
display(df_right)
print("\nMerged DataFrame (Inner Join):")
df_merged = pd.merge(df_left, df_right, on='Emp_ID')
display(df_merged)
```

### Task 2: Create two Pandas DataFrames with different columns. Concatenate the DataFrames along the rows and along the columns.

```python
df_c1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df_c2 = pd.DataFrame({'C': [5, 6], 'D': [7, 8]})
print("DataFrame 1:")
display(df_c1)
print("\nDataFrame 2:")
display(df_c2)
print("\nConcatenation along Rows (axis=0):")
display(pd.concat([df_c1, df_c2], axis=0, sort=False))
print("\nConcatenation along Columns (axis=1):")
display(pd.concat([df_c1, df_c2], axis=1))
```

# Assignment 6: Time Series Analysis

### Task 1: Create a Pandas DataFrame with a datetime index and one column filled with random integers. Resample the DataFrame to compute the monthly mean of the values.

```python
np.random.seed(42)
dates_daily = pd.date_range(start='2023-01-01', end='2023-06-30', freq='D')
df_ts1 = pd.DataFrame(np.random.randint(10, 100, size=len(dates_daily)), index=dates_daily, columns=['Value'])
print("First 5 rows of daily data:")
display(df_ts1.head())
print("\nMonthly Mean Resampled:")
df_monthly = df_ts1.resample('M').mean()
display(df_monthly)
```

### Task 2: Create a Pandas DataFrame with a datetime index ranging from '2021-01-01' to '2021-12-31' and one column filled with random integers. Compute the rolling mean with a window of 7 days.

```python
np.random.seed(42)
dates_2021 = pd.date_range(start='2021-01-01', end='2021-12-31', freq='D')
df_ts2 = pd.DataFrame(np.random.randint(10, 100, size=len(dates_2021)), index=dates_2021, columns=['Value'])
df_ts2['7D_Rolling_Mean'] = df_ts2['Value'].rolling(window=7).mean()
print("Showing head of the rolling calculations (first 10 days):")
display(df_ts2.head(10))
```

# Assignment 7: MultiIndex DataFrame

### Task 1: Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform some basic indexing and slicing operations on the MultiIndex DataFrame.

```python
arrays = [
    ['Group1', 'Group1', 'Group2', 'Group2'],
    ['SubA', 'SubB', 'SubA', 'SubB']
]
index = pd.MultiIndex.from_arrays(arrays, names=('Main', 'Sub'))
df_mi1 = pd.DataFrame({'Value': [10, 20, 30, 40]}, index=index)
print("MultiIndex DataFrame:")
display(df_mi1)
print("\nIndexing 'Group1':")
display(df_mi1.loc['Group1'])
print("\nSlicing 'Group1' to 'Group2' for SubA:")
display(df_mi1.xs('SubA', level='Sub'))
```

### Task 2: Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.

```python
np.random.seed(42)
categories = ['Furniture', 'Furniture', 'Office Supplies', 'Office Supplies']
subcategories = ['Chairs', 'Tables', 'Paper', 'Pens']
mi = pd.MultiIndex.from_arrays([categories, subcategories], names=['Category', 'SubCategory'])
df_mi2 = pd.DataFrame({'Sales': np.random.randint(50, 500, size=4)}, index=mi)
print("MultiIndex Sales DataFrame:")
display(df_mi2)
print("\nSum of values aggregated by Category:")
display(df_mi2.groupby(level='Category').sum())
```

# Assignment 8: Pivot Tables

### Task 1: Create a Pandas DataFrame with columns 'Date', 'Category', and 'Value'. Create a pivot table to compute the sum of 'Value' for each 'Category' by 'Date'.

```python
data_pivot1 = {
    'Date': ['2023-10-01', '2023-10-01', '2023-10-02', '2023-10-02', '2023-10-01'],
    'Category': ['A', 'B', 'A', 'B', 'A'],
    'Value': [10, 20, 30, 40, 15]
}
df_p1 = pd.DataFrame(data_pivot1)
print("Original DataFrame:")
display(df_p1)
pivot_table1 = df_p1.pivot_table(values='Value', index='Date', columns='Category', aggfunc='sum')
print("\nPivot Table (Sum of Value):")
display(pivot_table1)
```

### Task 2: Create a Pandas DataFrame with columns 'Year', 'Quarter', and 'Revenue'. Create a pivot table to compute the mean 'Revenue' for each 'Quarter' by 'Year'.

```python
data_pivot2 = {
    'Year': [2021, 2021, 2021, 2022, 2022, 2022],
    'Quarter': ['Q1', 'Q2', 'Q1', 'Q1', 'Q2', 'Q2'],
    'Revenue': [100, 150, 120, 200, 250, 270]
}
df_p2 = pd.DataFrame(data_pivot2)
print("Original DataFrame:")
display(df_p2)
pivot_table2 = df_p2.pivot_table(values='Revenue', index='Year', columns='Quarter', aggfunc='mean')
print("\nPivot Table (Mean Revenue):")
display(pivot_table2)
```

# Assignment 9: Applying Functions

### Task 1: Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Apply a function that doubles the values of the DataFrame.

```python
np.random.seed(42)
df_apply1 = pd.DataFrame(np.random.randint(1, 20, size=(5, 3)), columns=['A', 'B', 'C'])
print("Original DataFrame:")
display(df_apply1)
print("\nDataFrame with values doubled:")
double_func = lambda x: x * 2
display(df_apply1.apply(double_func))
```

### Task 2: Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers. Apply a lambda function to create a new column that is the sum of the existing columns.

```python
np.random.seed(42)
df_apply2 = pd.DataFrame(np.random.randint(1, 20, size=(6, 3)), columns=['X', 'Y', 'Z'])
df_apply2['Row_Sum'] = df_apply2.apply(lambda row: row['X'] + row['Y'] + row['Z'], axis=1)
print("DataFrame with row-wise sum column:")
display(df_apply2)
```
