```python
import pandas as pd
import numpy as np

# Create random categories
categories = np.random.choice(['A', 'B', 'C'], size=10)

# Create random values
values = np.random.randint(1, 100, size=10)

# Create DataFrame
df = pd.DataFrame({
    'Category': categories,
    'Value': values
})

print("DataFrame:")
print(df)

# Group by Category and calculate sum and mean
result = df.groupby('Category')['Value'].agg(['sum', 'mean'])

print("\nSum and Mean for Each Category:")
print(result)
```

```python
import pandas as pd

# Create first DataFrame
df1 = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4, 5, 6]
})

# Create second DataFrame
df2 = pd.DataFrame({
    'A': [7, 8, 9],
    'B': [10, 11, 12]
})

print("DataFrame 1:")
print(df1)

print("\nDataFrame 2:")
print(df2)

# Concatenate along rows
row_concat = pd.concat([df1, df2], axis=0, ignore_index=True)

print("\nConcatenated Along Rows:")
print(row_concat)

# Create DataFrames for column-wise concatenation
df3 = pd.DataFrame({
    'C': [13, 14, 15],
    'D': [16, 17, 18]
})

# Concatenate along columns
column_concat = pd.concat([df1, df3], axis=1)

print("\nConcatenated Along Columns:")
print(column_concat)
```

```python
import pandas as pd
import numpy as np

# Create a datetime index
dates = pd.date_range(
    start='2021-01-01',
    end='2021-12-31',
    freq='D'
)

# Create random values
values = np.random.randint(1, 100, size=len(dates))

# Create DataFrame
df = pd.DataFrame({
    'Value': values
}, index=dates)

print("Original DataFrame:")
print(df)

# Resample by month and calculate mean
monthly_mean = df.resample('ME')['Value'].mean()

print("\nMonthly Mean:")
print(monthly_mean)
```

```python
import pandas as pd
import numpy as np

# Create date range from 2021-01-01 to 2021-12-31
dates = pd.date_range(
    start='2021-01-01',
    end='2021-12-31',
    freq='D'
)

# Create random integers
values = np.random.randint(1, 100, size=len(dates))

# Create DataFrame
df = pd.DataFrame({
    'Value': values
}, index=dates)

print("Original DataFrame:")
print(df)

# Calculate 7-day rolling mean
rolling_mean = df['Value'].rolling(window=7).mean()

print("\n7-Day Rolling Mean:")
print(rolling_mean)
```

```python
import pandas as pd
import numpy as np

# Create MultiIndex
index = pd.MultiIndex.from_tuples([
    ('A', 1),
    ('A', 2),
    ('A', 3),
    ('B', 1),
    ('B', 2),
    ('B', 3)
], names=['Category', 'Number'])

# Create DataFrame
df = pd.DataFrame(
    {'Value': np.random.randint(1, 100, size=6)},
    index=index
)

print("MultiIndex DataFrame:")
print(df)

# Basic indexing
print("\nData for Category A:")
print(df.loc['A'])

# Access a specific row
print("\nValue at (A, 2):")
print(df.loc[('A', 2)])

# Slicing
print("\nRows from A to B:")
print(df.loc['A':'B'])
```

```python
import pandas as pd
import numpy as np

# Create categories and subcategories
categories = ['A', 'A', 'A', 'B', 'B', 'B']
subcategories = ['X', 'Y', 'Z', 'X', 'Y', 'Z']

# Create MultiIndex
index = pd.MultiIndex.from_arrays(
    [categories, subcategories],
    names=['Category', 'SubCategory']
)

# Create DataFrame
df = pd.DataFrame(
    {'Value': np.random.randint(1, 100, size=6)},
    index=index
)

print("MultiIndex DataFrame:")
print(df)

# Sum for each Category and SubCategory
result = df.groupby(
    level=['Category', 'SubCategory']
).sum()

print("\nSum of Values for Each Category and SubCategory:")
print(result)
```

```python
import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame({
    'Date': pd.to_datetime([
        '2021-01-01',
        '2021-01-01',
        '2021-01-02',
        '2021-01-02',
        '2021-01-03',
        '2021-01-03'
    ]),
    'Category': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Value': np.random.randint(10, 100, size=6)
})

print("Original DataFrame:")
print(df)

# Create pivot table
pivot = pd.pivot_table(
    df,
    values='Value',
    index='Date',
    columns='Category',
    aggfunc='sum'
)

print("\nPivot Table:")
print(pivot)
```

```python
import pandas as pd
import numpy as np

# Create a 6x3 DataFrame
data = np.random.randint(1, 50, size=(6, 3))

df = pd.DataFrame(
    data,
    columns=['A', 'B', 'C']
)

print("Original DataFrame:")
print(df)

# Create a new column containing the sum of A, B, and C
df['Sum'] = df.apply(
    lambda row: row['A'] + row['B'] + row['C'],
    axis=1
)

print("\nDataFrame with Sum Column:")
print(df)
```
