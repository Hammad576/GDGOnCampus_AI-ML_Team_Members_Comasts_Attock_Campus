### Assignment 2: Array Indexing and Slicing

**Task 1:** Create a NumPy array of shape (6, 6) with values from 1 to 36. Extract the sub-array consisting of the 3rd to 5th rows and 2nd to 4th columns.

```python
arr_2_1 = np.arange(1, 37).reshape(6, 6)
print("Original Array:")
print(arr_2_1)

# Extraction (3rd to 5th rows -> index 2:5, 2nd to 4th columns -> index 1:4)
sub_arr = arr_2_1[2:5, 1:4]
print("\nExtracted Sub-array:")
print(sub_arr)
```

**Task 2:** Create a NumPy array of shape (5, 5) with random integers. Extract the elements on the border.

```python
arr_2_2 = np.random.randint(1, 100, size=(5, 5))
print("Original Array:")
print(arr_2_2)

# Border elements extraction
# Top row, bottom row, left column (excluding corners), right column (excluding corners)
border_elements = np.concatenate([
    arr_2_2[0, :],          # Top row
    arr_2_2[-1, :],         # Bottom row
    arr_2_2[1:-1, 0],       # Left border (excluding top/bottom corner)
    arr_2_2[1:-1, -1]       # Right border (excluding top/bottom corner)
])
print("\nBorder Elements:")
print(border_elements)
```

### Assignment 3: Array Operations

### Assignment 4: Statistical Operations

**Task 1:** Create a NumPy array of shape (5, 5) filled with random integers. Compute the mean, median, standard deviation, and variance of the array.

```python
arr_4_1 = np.random.randint(1, 100, size=(5, 5))
print("Array:")
print(arr_4_1)

print("\nMean:", np.mean(arr_4_1))
print("Median:", np.median(arr_4_1))
print("Standard Deviation:", np.std(arr_4_1))
print("Variance:", np.var(arr_4_1))
```

**Task 2:** Create a NumPy array of shape (3, 3) with values from 1 to 9. Normalize the array (i.e., scale the values to have a mean of 0 and a standard deviation of 1).

```python
arr_4_2 = np.arange(1, 10).reshape(3, 3).astype(float)
print("Original Array:")
print(arr_4_2)

mean = np.mean(arr_4_2)
std = np.std(arr_4_2)

normalized_arr = (arr_4_2 - mean) / std
print("\nNormalized Array:")
print(normalized_arr)
```

### Assignment 5: Broadcasting

**Task 1:** Create a NumPy array of shape (3, 3) filled with random integers. Add a 1D array of shape (3,) to each row of the 2D array using broadcasting.

```python
arr_2d_1 = np.random.randint(1, 10, size=(3, 3))
arr_1d_1 = np.array([10, 20, 30])

print("2D Array:")
print(arr_2d_1)
print("\n1D Array:")
print(arr_1d_1)

result_5_1 = arr_2d_1 + arr_1d_1
print("\nResult after broadcasting (addition):")
print(result_5_1)
```

**Task 2:** Create a NumPy array of shape (4, 4) filled with random integers. Subtract a 1D array of shape (4,) from each column of the 2D array using broadcasting.

```python
arr_2d_2 = np.random.randint(1, 10, size=(4, 4))
arr_1d_2 = np.array([1, 2, 3, 4])

print("2D Array:")
print(arr_2d_2)
print("\n1D Array:")
print(arr_1d_2)

# Reshape 1D array to (4, 1) to enable column-wise broadcasting
result_5_2 = arr_2d_2 - arr_1d_2.reshape(4, 1)
print("\nResult after broadcasting (subtraction column-wise):")
print(result_5_2)
```

### Assignment 6: Linear Algebra

**Task 1:** Create a NumPy array of shape (3, 3) representing a matrix. Compute its determinant, inverse, and eigenvalues.

```python
matrix_a = np.array([[2, 1, 1], [1, 3, 2], [1, 0, 1]], dtype=float)
print("Matrix:")
print(matrix_a)

det = np.linalg.det(matrix_a)
inv = np.linalg.inv(matrix_a)
eigenvalues, _ = np.linalg.eig(matrix_a)

print(f"\nDeterminant: {det}")
print("\nInverse Matrix:")
print(inv)
print("\nEigenvalues:")
print(eigenvalues)
```

**Task 2:** Create two NumPy arrays of shape (2, 3) and (3, 2). Perform matrix multiplication on these arrays.

### Assignment 9: Structured Arrays

**Task 1:** Create a structured array with fields 'name' (string), 'age' (integer), and 'weight' (float). Add some data and sort the array by age.

```python
dtype_9_1 = [('name', 'U20'), ('age', 'i4'), ('weight', 'f4')]
data_9_1 = [
    ('Alice', 25, 55.5),
    ('Bob', 20, 72.0),
    ('Charlie', 30, 68.2)
]

structured_arr = np.array(data_9_1, dtype=dtype_9_1)
print("Original Structured Array:")
print(structured_arr)

sorted_arr = np.sort(structured_arr, order='age')
print("\nSorted by Age:")
print(sorted_arr)
```

**Task 2:** Create a structured array with fields 'x' and 'y' (both integers). Add some data and compute the Euclidean distance between each pair of points.

```python
dtype_9_2 = [('x', 'i4'), ('y', 'i4')]
points_data = [(1, 2), (4, 6), (7, 10)]
points = np.array(points_data, dtype=dtype_9_2)

print("Points:")
print(points)

# Compute pairwise Euclidean distances
n_points = len(points)
print("\nPairwise Euclidean Distances:")
for i in range(n_points):
    for j in range(i + 1, n_points):
        p1 = points[i]
        p2 = points[j]
        distance = np.sqrt((p1['x'] - p2['x'])**2 + (p1['y'] - p2['y'])**2)
        print(f"Distance between Point {i} {p1} and Point {j} {p2}: {distance:.2f}")
```

### Assignment 10: Masked Arrays

**Task 1:** Create a masked array of shape (4, 4) with random integers and mask the elements greater than 10. Compute the sum of the unmasked elements.

```python
import numpy.ma as ma

raw_data_10 = np.random.randint(1, 20, size=(4, 4))
print("Raw Data:")
print(raw_data_10)

masked_arr_1 = ma.masked_greater(raw_data_10, 10)
print("\nMasked Array (elements > 10 masked):")
print(masked_arr_1)

print("\nSum of Unmasked Elements:", masked_arr_1.sum())
```

**Task 2:** Create a masked array of shape (3, 3) with random integers and mask the diagonal elements. Replace the masked elements with the mean of the unmasked elements.

```python
raw_data_10_2 = np.random.randint(1, 20, size=(3, 3))
print("Raw Data:")
print(raw_data_10_2)

# Create a mask for the diagonal elements
diagonal_mask = np.eye(3, dtype=bool)
masked_arr_2 = ma.array(raw_data_10_2, mask=diagonal_mask)
print("\nMasked Array (diagonal masked):")
print(masked_arr_2)

mean_val = masked_arr_2.mean()
print(f"\nMean of unmasked elements: {mean_val:.2f}")

filled_arr = masked_arr_2.filled(mean_val)
print("\nArray with masked diagonal elements replaced by mean:")
print(filled_arr)
```
