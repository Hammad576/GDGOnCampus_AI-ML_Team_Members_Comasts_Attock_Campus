```python
import numpy as np

# Create a 3x3 matrix
matrix = np.array([
    [2, 1, 3],
    [1, 2, 1],
    [3, 1, 2]
])

print("Matrix:")
print(matrix)

# Calculate determinant
determinant = np.linalg.det(matrix)

print("\nDeterminant:")
print(determinant)

# Calculate inverse
inverse = np.linalg.inv(matrix)

print("\nInverse:")
print(inverse)

# Calculate eigenvalues
eigenvalues = np.linalg.eigvals(matrix)

print("\nEigenvalues:")
print(eigenvalues)
```

QUESTIOM_2:

```python
import numpy as np

# Create a 3x3 array with random integers
arr = np.random.randint(1, 20, size=(3, 3))

print("Original Array:")
print(arr)

# Create a mask for the diagonal elements
mask = np.eye(3, dtype=bool)

# Create the masked array
masked_arr = np.ma.array(arr, mask=mask)

print("\nMasked Array:")
print(masked_arr)

# Calculate the mean of unmasked elements
mean_value = masked_arr.mean()

print("\nMean of Unmasked Elements:")
print(mean_value)

# Fill the masked elements with the mean
result = masked_arr.filled(mean_value)

print("\nArray After Replacing Diagonal Elements:")
print(result)
```
