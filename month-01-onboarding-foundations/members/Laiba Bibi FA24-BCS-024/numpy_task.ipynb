```python
# Q1
dtype = [('name', 'U10'), ('age', 'i4'), ('weight', 'f4')]
people = np.array([('Ali', 25, 70.5),
                   ('Sara', 22, 55.2),
                   ('Ahmed', 30, 82.1),
                   ('Fatima', 19, 60.0)], dtype=dtype)
print(np.sort(people, order='age'))
```

```python
# Q2
points = np.array([(0, 0), (3, 4), (6, 8), (1, 1)],
                  dtype=[('x', 'i4'), ('y', 'i4')])
x = points['x']
y = points['y']

dist = np.sqrt((x[:, None] - x[None, :])**2 + (y[:, None] - y[None, :])**2)
print(np.round(dist, 3))

n = len(points)
for i in range(n):
    for j in range(i + 1, n):
        print(f"Point {i} to Point {j}: {dist[i, j]:.3f}")
```

```python
# Q2
data = np.random.randint(1, 20, size=(3, 3))
mask = np.eye(3, dtype=bool)
masked = np.ma.masked_array(data, mask=mask)
print(masked)

mean_val = masked.mean()
print("Mean of unmasked:", mean_val)
print(masked.astype(float).filled(mean_val))
```
