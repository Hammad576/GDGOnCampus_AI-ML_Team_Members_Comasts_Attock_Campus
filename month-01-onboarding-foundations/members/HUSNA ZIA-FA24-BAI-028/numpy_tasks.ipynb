```python
import numpy as np

# Q1
arr = np.random.randint(1, 21, size=(5, 5))

print("Original Array:")
print(arr)

arr[:, 2] = 1

print("After replacing third column with 1:")
print(arr)


# Q2
arr = np.arange(1, 17).reshape(4, 4)

print("\nOriginal Array:")
print(arr)

np.fill_diagonal(arr, 0)

print("After replacing diagonal with 0:")
print(arr)
```

```python
import numpy as np

# Q1
arr = np.arange(1, 37).reshape(6, 6)

print("Original Array:")
print(arr)

sub_array = arr[2:5, 1:4]

print("Sub-array:")
print(sub_array)


# Q2
arr = np.random.randint(1, 21, size=(5, 5))

print("\nOriginal Array:")
print(arr)

border = np.concatenate([
    arr[0, :],
    arr[-1, :],
    arr[1:-1, 0],
    arr[1:-1, -1]
])

print("Border Elements:")
print(border)
```

```python
import numpy as np

# Q1
arr1 = np.random.randint(1, 11, size=(3, 4))
arr2 = np.random.randint(1, 11, size=(3, 4))

print("Array 1:")
print(arr1)

print("Array 2:")
print(arr2)

print("Addition:")
print(arr1 + arr2)

print("Subtraction:")
print(arr1 - arr2)

print("Multiplication:")
print(arr1 * arr2)

print("Division:")
print(arr1 / arr2)


# Q2
arr = np.arange(1, 17).reshape(4, 4)

print("\nArray:")
print(arr)

print("Row-wise Sum:")
print(np.sum(arr, axis=1))

print("Column-wise Sum:")
print(np.sum(arr, axis=0))
```

```python
import numpy as np

# Q1
arr = np.random.randint(1, 21, size=(5, 5))

print("Array:")
print(arr)

print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:", np.std(arr))
print("Variance:", np.var(arr))


# Q2
arr = np.arange(1, 10).reshape(3, 3)

print("\nOriginal Array:")
print(arr)

normalized = (arr - np.mean(arr)) / np.std(arr)

print("Normalized Array:")
print(normalized)
```

```python
import numpy as np

# Q1
arr = np.random.randint(1, 11, size=(3, 3))
one_d = np.array([10, 20, 30])

print("2D Array:")
print(arr)

print("1D Array:")
print(one_d)

print("After Broadcasting:")
print(arr + one_d)


# Q2
arr = np.random.randint(1, 21, size=(4, 4))
one_d = np.array([1, 2, 3, 4])

print("\n2D Array:")
print(arr)

print("1D Array:")
print(one_d)

print("After Broadcasting:")
print(arr - one_d.reshape(4, 1))
```

```python
import numpy as np

# Q1
matrix = np.array([
    [1, 2, 3],
    [0, 1, 4],
    [5, 6, 0]
])

print("Matrix:")
print(matrix)

print("Determinant:")
print(np.linalg.det(matrix))

print("Inverse:")
print(np.linalg.inv(matrix))

print("Eigenvalues:")
print(np.linalg.eigvals(matrix))


# Q2
arr1 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

arr2 = np.array([
    [7, 8],
    [9, 10],
    [11, 12]
])

print("\nMatrix Multiplication:")
print(np.matmul(arr1, arr2))
```

```python
import numpy as np

# Q1
data = np.array([
    ("Ali", 22, 65.5),
    ("Husna", 20, 55.2),
    ("Sara", 24, 60.8),
    ("Ahmed", 21, 70.1)
], dtype=[
    ("name", "U20"),
    ("age", "i4"),
    ("weight", "f4")
])

print("Original Data:")
print(data)

print("\nSorted by Age:")
print(np.sort(data, order="age"))


# Q2
points = np.array([
    (1, 2),
    (3, 4),
    (5, 6),
    (7, 8)
], dtype=[
    ("x", "i4"),
    ("y", "i4")
])

print("\nPoints:")
print(points)

distance = np.sqrt(points["x"] ** 2 + points["y"] ** 2)

print("Euclidean Distance:")
print(distance)
```

```python
import numpy as np

# Q1
arr = np.random.randint(1, 21, size=(4, 4))

masked_arr = np.ma.masked_greater(arr, 10)

print("Original Array:")
print(arr)

print("Masked Array:")
print(masked_arr)

print("Sum of Unmasked Elements:")
print(np.ma.sum(masked_arr))


# Q2
arr = np.random.randint(1, 21, size=(3, 3))

mask = np.eye(3, dtype=bool)

masked_arr = np.ma.array(arr, mask=mask)

print("\nOriginal Array:")
print(arr)

print("Masked Diagonal:")
print(masked_arr)

mean_value = masked_arr.mean()

result = masked_arr.filled(mean_value)

print("After replacing diagonal with mean:")
print(result)
```
