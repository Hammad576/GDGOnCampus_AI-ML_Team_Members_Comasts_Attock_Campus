### Assignment 3: Data Cleaning

**Q1.** NaN values daalo aur har column ke mean se fill karo.

```python
df = pd.DataFrame(np.random.randint(1, 50, size=(5, 3)), columns=['A', 'B', 'C']).astype(float)

# NaN introduce karna
df.iloc[0, 1] = np.nan
df.iloc[2, 0] = np.nan
df.iloc[3, 2] = np.nan
df.iloc[4, 1] = np.nan
print("With NaN:\n", df)

print("\nColumn means (NaN ignore hote hain):\n", df.mean().round(2))

df_filled = df.fillna(df.mean())
print("\nNaN filled with column mean:\n", df_filled.round(2))
```

**Q2.** 4 columns, 6 rows. NaN daalo aur woh rows drop karo jinme koi bhi NaN ho.

```python
df = pd.DataFrame(np.random.randint(1, 50, size=(6, 4)), columns=['A', 'B', 'C', 'D']).astype(float)

df.iloc[1, 2] = np.nan
df.iloc[3, 0] = np.nan
df.iloc[5, 3] = np.nan
print("With NaN:\n", df)

df_clean = df.dropna()
print("\nAfter dropping rows with any NaN:\n", df_clean)
```

### Assignment 4: Data Aggregation

**Q1.** 'Category' (A/B/C) aur 'Value' columns. Category ke hisaab se Value ka sum aur mean.

```python
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], size=10),
    'Value': np.random.randint(1, 100, size=10)
})
print("DataFrame:\n", df)

result = df.groupby('Category')['Value'].agg(['sum', 'mean'])
print("\nSum and mean per Category:\n", result)
```

**Q2.** 'Product', 'Category', 'Sales' columns. Category ke hisaab se total sales.

```python
df = pd.DataFrame({
    'Product':  [f'Product_{i}' for i in range(1, 11)],
    'Category': np.random.choice(['Electronics', 'Clothing', 'Grocery'], size=10),
    'Sales':    np.random.randint(100, 1000, size=10)
})
print("DataFrame:\n", df)

total_sales = df.groupby('Category')['Sales'].sum()
print("\nTotal sales per Category:\n", total_sales)
```

### Assignment 5: Merging DataFrames

**Q1.** Common column wale do DataFrames ko merge karo.

```python
df1 = pd.DataFrame({'ID': [1, 2, 3, 4], 'Name': ['Ali', 'Sara', 'Ahmed', 'Maryam']})
df2 = pd.DataFrame({'ID': [2, 3, 4, 5], 'Marks': [85, 90, 78, 88]})
print("df1:\n", df1, "\n\ndf2:\n", df2)

print("\nInner merge on 'ID':\n", pd.merge(df1, df2, on='ID', how='inner'))
print("\nOuter merge on 'ID':\n", pd.merge(df1, df2, on='ID', how='outer'))
```

**Q2.** Alag columns wale do DataFrames ko rows aur columns dono ke along concatenate karo.

```python
df1 = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
df2 = pd.DataFrame({'C': [7, 8, 9], 'D': [10, 11, 12]})
print("df1:\n", df1, "\n\ndf2:\n", df2)

print("\nConcat along rows (axis=0) - alag columns ki wajah se NaN aayenge:\n",
      pd.concat([df1, df2], axis=0, ignore_index=True))
print("\nConcat along columns (axis=1):\n", pd.concat([df1, df2], axis=1))
```

### Assignment 6: Time Series Analysis

**Q1.** Datetime index wala DataFrame banao aur monthly mean nikalo (resample).

```python
dates = pd.date_range(start='2021-01-01', periods=90, freq='D')
df = pd.DataFrame({'Value': np.random.randint(1, 100, size=90)}, index=dates)
print("First 5 rows:\n", df.head())

monthly_mean = df.resample('MS').mean()   # MS = month start (label har month ki pehli tareekh)
print("\nMonthly mean:\n", monthly_mean.round(2))
```

**Q2.** '2021-01-01' se '2021-12-31' tak ka index, 7-day rolling mean.

```python
dates = pd.date_range(start='2021-01-01', end='2021-12-31', freq='D')
df = pd.DataFrame({'Value': np.random.randint(1, 100, size=len(dates))}, index=dates)

df['Rolling_Mean_7d'] = df['Value'].rolling(window=7).mean()
print("Total days:", len(df))
print(df.head(10).round(2))
print("\n(Pehle 6 din NaN hain kyunke 7 values chahiye window poori karne ke liye.)")
```

### Assignment 7: MultiIndex DataFrame

**Q1.** MultiIndex DataFrame banao aur basic indexing / slicing karo.

```python
index = pd.MultiIndex.from_product([['Group1', 'Group2'], [2021, 2022, 2023]],
                                   names=['Group', 'Year'])
df = pd.DataFrame({'Sales': np.random.randint(100, 500, 6),
                   'Profit': np.random.randint(10, 100, 6)}, index=index)
print("MultiIndex DataFrame:\n", df)

print("\n1) Outer level select: df.loc['Group1']\n", df.loc['Group1'])
print("\n2) Specific row: df.loc[('Group2', 2022)]\n", df.loc[('Group2', 2022)])
print("\n3) Cross-section (Year=2022): df.xs(2022, level='Year')\n", df.xs(2022, level='Year'))
print("\n4) Slice (saare groups, Year 2022-2023):\n", df.loc[pd.IndexSlice[:, 2022:2023], :])
print("\n5) Single column: df['Sales']\n", df['Sales'])
```

**Q2.** MultiIndex ('Category', 'SubCategory'), random data, har Category aur SubCategory ka sum.

```python
tuples = [('Electronics', 'Phone'), ('Electronics', 'Phone'), ('Electronics', 'Laptop'),
          ('Electronics', 'Laptop'), ('Clothing', 'Shirt'), ('Clothing', 'Shirt'),
          ('Clothing', 'Jeans'), ('Clothing', 'Jeans')]
index = pd.MultiIndex.from_tuples(tuples, names=['Category', 'SubCategory'])
df = pd.DataFrame({'Value': np.random.randint(10, 100, size=len(tuples))}, index=index)
print("DataFrame:\n", df)

print("\nSum per Category & SubCategory:\n", df.groupby(level=['Category', 'SubCategory']).sum())
print("\nSum per Category:\n", df.groupby(level='Category').sum())
```

### Assignment 8: Pivot Tables

**Q1.** 'Date', 'Category', 'Value'. Pivot table: har Date par har Category ka Value sum.

```python
df = pd.DataFrame({
    'Date': np.repeat(pd.date_range('2021-01-01', periods=4), 3),
    'Category': ['A', 'B', 'C'] * 4,
    'Value': np.random.randint(10, 100, size=12)
})
print("DataFrame:\n", df)

pivot = df.pivot_table(values='Value', index='Date', columns='Category', aggfunc='sum')
print("\nPivot table (sum of Value):\n", pivot)
```

**Q2.** 'Year', 'Quarter', 'Revenue'. Pivot table: har Year me har Quarter ka mean Revenue.

```python
df = pd.DataFrame({
    'Year': np.repeat([2021, 2022, 2023], 8),
    'Quarter': ['Q1', 'Q1', 'Q2', 'Q2', 'Q3', 'Q3', 'Q4', 'Q4'] * 3,
    'Revenue': np.random.randint(1000, 5000, size=24)
})
print("First 8 rows:\n", df.head(8))

pivot = df.pivot_table(values='Revenue', index='Year', columns='Quarter', aggfunc='mean')
print("\nPivot table (mean Revenue):\n", pivot.round(2))
```

### Assignment 9: Applying Functions

**Q1.** 3 columns, 5 rows. Ek function apply karo jo values ko double kare.

```python
df = pd.DataFrame(np.random.randint(1, 20, size=(5, 3)), columns=['A', 'B', 'C'])
print("Original:\n", df)

def double(x):
    return x * 2

doubled = df.apply(double)     # har column par apply hota hai
print("\nDoubled:\n", doubled)
```

**Q2.** 3 columns, 6 rows. Lambda function se naya column jo existing columns ka sum ho.

```python
df = pd.DataFrame(np.random.randint(1, 20, size=(6, 3)), columns=['A', 'B', 'C'])

df['Sum'] = df.apply(lambda row: row['A'] + row['B'] + row['C'], axis=1)
print(df)
```

### Assignment 10: Working with Text Data

**Q1.** 5 random text strings ki Series, sab ko uppercase karo.

```python
def random_text(n=8):
    letters = list('abcdefghijklmnopqrstuvwxyz')
    return ''.join(np.random.choice(letters, size=n))

s = pd.Series([random_text() for _ in range(5)])
print("Original:\n", s)

print("\nUppercase:\n", s.str.upper())
```

**Q2.** 5 random text strings ki Series, har string ke pehle teen characters extract karo.

```python
s = pd.Series([random_text() for _ in range(5)])
print("Original:\n", s)

print("\nFirst three characters:\n", s.str[:3])      # ya s.str.slice(0, 3)
```
