```python
import pandas as pd
import numpy as np

# 1. 4 columns, 6 rows
df = pd.DataFrame(np.random.randint(1, 50, (6, 4)),
                  columns=['A', 'B', 'C', 'D'])

df = df.set_index('A')
print(df)


# 2. Access row Y, column B
df = pd.DataFrame(np.random.randint(1, 50, (3, 3)),
                  index=['X', 'Y', 'Z'],
                  columns=['A', 'B', 'C'])

print(df)
print("Element:", df.loc['Y', 'B'])
```

```python
import pandas as pd
import numpy as np

# 1. Product of first two columns
df = pd.DataFrame(np.random.randint(1, 10, (5, 3)),
                  columns=['A', 'B', 'C'])

df['Product'] = df['A'] * df['B']
print(df)


# 2. Row-wise and column-wise sum
df = pd.DataFrame(np.random.randint(1, 10, (4, 3)),
                  columns=['A', 'B', 'C'])

print("Row-wise sum:")
print(df.sum(axis=1))

print("Column-wise sum:")
print(df.sum(axis=0))
```

```python
import pandas as pd
import numpy as np

# 1. Fill NaN with column mean
df = pd.DataFrame(np.random.randint(1, 50, (5, 3)),
                  columns=['A', 'B', 'C'])

df.loc[1, 'A'] = np.nan
df.loc[3, 'B'] = np.nan
df.loc[2, 'C'] = np.nan

print("Before:")
print(df)

df = df.fillna(df.mean())

print("After:")
print(df)


# 2. Drop rows containing NaN
df = pd.DataFrame(np.random.randint(1, 50, (6, 4)),
                  columns=['A', 'B', 'C', 'D'])

df.loc[1, 'A'] = np.nan
df.loc[4, 'C'] = np.nan

print("Before:")
print(df)

df = df.dropna()

print("After:")
print(df)
```

```python
import pandas as pd
import numpy as np

# 1. Group by Category
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], 10),
    'Value': np.random.randint(10, 100, 10)
})

result = df.groupby('Category')['Value'].agg(['sum', 'mean'])
print(result)


# 2. Total sales by Category
df = pd.DataFrame({
    'Product': np.random.choice(['Laptop', 'Phone', 'Tablet'], 10),
    'Category': np.random.choice(['A', 'B', 'C'], 10),
    'Sales': np.random.randint(100, 1000, 10)
})

print(df.groupby('Category')['Sales'].sum())
```

```python
import pandas as pd

# 1. Merge using common column
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Ali', 'Hassan', 'Sara']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [80, 75, 90]
})

merged = pd.merge(df1, df2, on='ID')
print(merged)


# 2. Concatenate rows and columns
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'C': [5, 6], 'D': [7, 8]})

print("Row-wise:")
print(pd.concat([df1, df2], axis=0, ignore_index=True))

print("Column-wise:")
print(pd.concat([df1, df2], axis=1))
```

```python
import pandas as pd
import numpy as np

# 1. Monthly mean
dates = pd.date_range('2021-01-01', periods=365)

df = pd.DataFrame({
    'Value': np.random.randint(1, 100, 365)
}, index=dates)

print(df.resample('ME').mean())


# 2. 7-day rolling mean
dates = pd.date_range('2021-01-01', '2021-12-31')

df = pd.DataFrame({
    'Value': np.random.randint(1, 100, len(dates))
}, index=dates)

df['Rolling_Mean'] = df['Value'].rolling(7).mean()

print(df)
```

```python
import pandas as pd
import numpy as np

# 1. MultiIndex
index = pd.MultiIndex.from_tuples([
    ('A', 1), ('A', 2), ('B', 1), ('B', 2)
])

df = pd.DataFrame({
    'Value': [10, 20, 30, 40]
}, index=index)

print(df)

print("Category A:")
print(df.loc['A'])

print("A, row 2:")
print(df.loc[('A', 2)])


# 2. Category and SubCategory
index = pd.MultiIndex.from_product([
    ['A', 'B', 'C'],
    ['X', 'Y']
], names=['Category', 'SubCategory'])

df = pd.DataFrame({
    'Value': np.random.randint(1, 50, 6)
}, index=index)

print(df)

print("\nSum:")
print(df.groupby(level=['Category', 'SubCategory']).sum())
```

```python
import pandas as pd
import numpy as np

# 1. Pivot table
df = pd.DataFrame({
    'Date': pd.date_range('2021-01-01', periods=6),
    'Category': ['A', 'B', 'A', 'B', 'C', 'C'],
    'Value': np.random.randint(10, 100, 6)
})

pivot = pd.pivot_table(
    df,
    values='Value',
    index='Date',
    columns='Category',
    aggfunc='sum'
)

print(pivot)


# 2. Revenue pivot table
df = pd.DataFrame({
    'Year': [2021, 2021, 2022, 2022, 2023, 2023],
    'Quarter': ['Q1', 'Q2', 'Q1', 'Q2', 'Q1', 'Q2'],
    'Revenue': np.random.randint(1000, 5000, 6)
})

pivot = pd.pivot_table(
    df,
    values='Revenue',
    index='Year',
    columns='Quarter',
    aggfunc='mean'
)

print(pivot)
```

```python
import pandas as pd
import numpy as np

# 1. Double all values
df = pd.DataFrame(
    np.random.randint(1, 20, (5, 3)),
    columns=['A', 'B', 'C']
)

df = df.map(lambda x: x * 2)

print(df)


# 2. Sum of existing columns
df = pd.DataFrame(
    np.random.randint(1, 20, (6, 3)),
    columns=['A', 'B', 'C']
)

df['Sum'] = df.apply(lambda row: row['A'] + row['B'] + row['C'], axis=1)

print(df)
```
