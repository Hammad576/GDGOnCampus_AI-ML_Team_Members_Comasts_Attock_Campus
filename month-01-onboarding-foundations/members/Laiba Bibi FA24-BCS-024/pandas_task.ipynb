```python
import pandas as pd
import numpy as np
np.random.seed(42)
```

Assignment 1: DataFrame Creation and Indexing

```python
# Q1: 4 columns, 6 rows, index set to the first column
df = pd.DataFrame(np.random.randint(1, 100, size=(6, 4)),
                  columns=['A', 'B', 'C', 'D'])
print("Original:\n", df)

df = df.set_index('A')
print("\nIndex set to column A:\n", df)
```

```python
# Q2: columns A, B, C and index X, Y, Z
df = pd.DataFrame(np.random.randint(1, 100, size=(3, 3)),
                  columns=['A', 'B', 'C'],
                  index=['X', 'Y', 'Z'])
print(df)
print("\nElement at row Y, column B:", df.loc['Y', 'B'])
```

Assignment 2: DataFrame Operations

```python
# Q1: new column = product of first two columns
df = pd.DataFrame(np.random.randint(1, 10, size=(5, 3)),
                  columns=['A', 'B', 'C'])
df['A_x_B'] = df['A'] * df['B']
print(df)
```

```python
# Q2: row-wise and column-wise sum
df = pd.DataFrame(np.random.randint(1, 10, size=(4, 3)),
                  columns=['A', 'B', 'C'])
print(df)
print("\nRow-wise sum:\n", df.sum(axis=1))
print("\nColumn-wise sum:\n", df.sum(axis=0))
```

Assignment 3: Data Cleaning

```python
# Q1: introduce NaN and fill with column mean
df = pd.DataFrame(np.random.randint(1, 50, size=(5, 3)),
                  columns=['A', 'B', 'C']).astype(float)
df.iloc[0, 1] = np.nan
df.iloc[2, 0] = np.nan
df.iloc[3, 2] = np.nan
print("With NaN:\n", df)

df_filled = df.fillna(df.mean())
print("\nFilled with column means:\n", df_filled)
```

```python
# Q2: introduce NaN and drop rows with any NaN
df = pd.DataFrame(np.random.randint(1, 50, size=(6, 4)),
                  columns=['A', 'B', 'C', 'D']).astype(float)
df.iloc[1, 2] = np.nan
df.iloc[4, 0] = np.nan
print("With NaN:\n", df)

df_clean = df.dropna()
print("\nAfter dropping rows with NaN:\n", df_clean)
```

Assignment 4: Data Aggregation

```python
# Q1: group by Category, sum and mean of Value
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], size=10),
    'Value': np.random.randint(1, 100, size=10)
})
print(df)

result = df.groupby('Category')['Value'].agg(['sum', 'mean'])
print("\nSum and mean per category:\n", result)
```

```python
# Q2: total sales per category
df = pd.DataFrame({
    'Product': ['P1', 'P2', 'P3', 'P4', 'P5', 'P6', 'P7', 'P8'],
    'Category': np.random.choice(['Electronics', 'Clothing', 'Food'], size=8),
    'Sales': np.random.randint(100, 1000, size=8)
})
print(df)

print("\nTotal sales per category:\n", df.groupby('Category')['Sales'].sum())
```

Assignment 5: Merging DataFrames

```python
# Q1: merge on a common column
df1 = pd.DataFrame({'ID': [1, 2, 3, 4],
                    'Name': ['Ali', 'Sara', 'Ahmed', 'Fatima']})
df2 = pd.DataFrame({'ID': [1, 2, 3, 5],
                    'Marks': [85, 90, 78, 88]})

merged = pd.merge(df1, df2, on='ID')  # inner join by default
print(merged)
```

```python
# Q2: concatenate along rows and along columns
df1 = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
df2 = pd.DataFrame({'C': [7, 8, 9], 'D': [10, 11, 12]})

print("Along rows (axis=0):\n", pd.concat([df1, df2], axis=0))
print("\nAlong columns (axis=1):\n", pd.concat([df1, df2], axis=1))
```

Assignment 6: Time Series Analysis

```python
# Q1: monthly mean using resample
dates = pd.date_range(start='2021-01-01', periods=100, freq='D')
df = pd.DataFrame({'Value': np.random.randint(1, 100, size=100)}, index=dates)

monthly_mean = df.resample('MS').mean()  # 'MS' = month start
print(monthly_mean)
```

```python
# Q2: rolling mean with a 7-day window
dates = pd.date_range(start='2021-01-01', end='2021-12-31', freq='D')
df = pd.DataFrame({'Value': np.random.randint(1, 100, size=len(dates))}, index=dates)

df['Rolling_Mean_7'] = df['Value'].rolling(window=7).mean()
print(df.head(10))  # first 6 rows are NaN because 7 values are needed
```

Assignment 7: MultiIndex DataFrame

```python
# Q1: MultiIndex, indexing and slicing
index = pd.MultiIndex.from_product([['Group1', 'Group2'], ['X', 'Y', 'Z']],
                                   names=['Group', 'Item'])
df = pd.DataFrame({'Value': np.random.randint(1, 100, size=6)}, index=index)
print(df)

print("\nAll rows of Group1:\n", df.loc['Group1'])
print("\nSingle row (Group2, Y):\n", df.loc[('Group2', 'Y')])
print("\nItem 'X' from every group:\n", df.xs('X', level='Item'))
```

```python
# Q2: Category and SubCategory, sum of values
index = pd.MultiIndex.from_tuples([
    ('Fruit', 'Apple'), ('Fruit', 'Banana'),
    ('Vegetable', 'Carrot'), ('Vegetable', 'Potato'),
    ('Fruit', 'Apple'), ('Vegetable', 'Carrot')
], names=['Category', 'SubCategory'])
df = pd.DataFrame({'Value': np.random.randint(1, 100, size=6)}, index=index)
print(df)

print("\nSum per Category and SubCategory:\n", df.groupby(level=['Category', 'SubCategory']).sum())
print("\nSum per Category:\n", df.groupby(level='Category').sum())
```

Assignment 8: Pivot Tables

```python
# Q1: sum of Value for each Category by Date
df = pd.DataFrame({
    'Date': pd.to_datetime(['2021-01-01', '2021-01-01', '2021-01-02',
                            '2021-01-02', '2021-01-03', '2021-01-03']),
    'Category': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Value': np.random.randint(1, 100, size=6)
})
print(df)

pivot = df.pivot_table(values='Value', index='Date', columns='Category', aggfunc='sum')
print("\nPivot table:\n", pivot)
```

```python
# Q2: mean Revenue for each Quarter by Year
df = pd.DataFrame({
    'Year': [2020, 2020, 2020, 2020, 2021, 2021, 2021, 2021],
    'Quarter': ['Q1', 'Q2', 'Q3', 'Q4', 'Q1', 'Q2', 'Q3', 'Q4'],
    'Revenue': np.random.randint(1000, 5000, size=8)
})
print(df)

pivot = df.pivot_table(values='Revenue', index='Year', columns='Quarter', aggfunc='mean')
print("\nPivot table:\n", pivot)
```

Assignment 9: Applying Functions

```python
# Q1: apply a function that doubles the values
df = pd.DataFrame(np.random.randint(1, 20, size=(5, 3)),
                  columns=['A', 'B', 'C'])
print("Original:\n", df)

def double(x):
    return x * 2

print("\nDoubled:\n", df.apply(double))
```

```python
# Q2: lambda to create a new column that is the sum of existing columns
df = pd.DataFrame(np.random.randint(1, 20, size=(6, 3)),
                  columns=['A', 'B', 'C'])
df['Sum'] = df.apply(lambda row: row['A'] + row['B'] + row['C'], axis=1)
print(df)
```
