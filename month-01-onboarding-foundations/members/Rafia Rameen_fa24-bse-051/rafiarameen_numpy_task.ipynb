**Assignment 01:** **Array Creation and manipulation**

```python
import numpy as np

# Task 1: 5x5 array
arr1_1 = np.random.randint(1, 21, size=(5, 5))
print("1. Original Array:\n", arr1_1)
arr1_1[:, 2] = 1
print("Replaced 3rd column with 1:\n", arr1_1)

# Task 2: 4x4 array
arr1_2 = np.arange(1, 17).reshape(4, 4)
print("\n2. Original Array:\n", arr1_2)
np.fill_diagonal(arr1_2, 0)
print("Diagonal elements set to 0:\n", arr1_2)
```

**Assignment 02: Array Indexing and Slicing**

```python
# Task 1:
arr2_1 = np.arange(1, 37).reshape(6, 6)
sub_arr = arr2_1[2:5, 1:4]
print("1. Extracted Sub-array:\n", sub_arr)

# Task 2:
arr2_2 = np.random.randint(1, 50, size=(5, 5))
print("\n2. Original 5x5 Array:\n", arr2_2)
top = arr2_2[0, :]
bottom = arr2_2[-1, :]
left = arr2_2[1:-1, 0]
right = arr2_2[1:-1, -1]
border_elements = np.concatenate([top, right, bottom[::-1], left[::-1]])
print("Border elements:", border_elements)
```

**Assignment 03: Array Operations**

```python
# Task 1:
a1 = np.random.randint(1, 10, size=(3, 4))
a2 = np.random.randint(1, 10, size=(3, 4))
print("1. Addition:\n", a1 + a2)
print("Subtraction:\n", a1 - a2)
print("Multiplication:\n", a1 * a2)
print("Division:\n", a1 / a2)

# Task 2:
arr3_2 = np.arange(1, 17).reshape(4, 4)
print("\n2. Row-wise sum:", np.sum(arr3_2, axis=1))
print("Column-wise sum:", np.sum(arr3_2, axis=0))
```

**Assignment 04: Statistical operations**

```python
# Task 1:
arr4_1 = np.random.randint(1, 100, size=(5, 5))
print("1. Mean:", np.mean(arr4_1))
print("Median:", np.median(arr4_1))
print("Standard Deviation:", np.std(arr4_1))
print("Variance:", np.var(arr4_1))

# Task 2:
arr4_2 = np.arange(1, 10).reshape(3, 3)
normalized = (arr4_2 - np.mean(arr4_2)) / np.std(arr4_2)
print("\n2. Normalized Array (Mean=0, Std=1):\n", normalized)
```

**Assignment 06: Linear Algebra**

```python
# Task 1:
matrix = np.array([[2, 1, 1], [1, 3, 2], [1, 0, 0]])
print("1. Determinant:", np.linalg.det(matrix))
print("Inverse:\n", np.linalg.inv(matrix))
eigenvalues, _ = np.linalg.eig(matrix)
print("Eigenvalues:", eigenvalues)

# Task 2:
m1 = np.random.randint(1, 10, size=(2, 3))
m2 = np.random.randint(1, 10, size=(3, 2))
print("\n2. Matrix Multiplication Result:\n", np.dot(m1, m2))
```

**Assignment 09: Structured Arrays**

```python
# Task 1:
dtype1 = [('name', 'S10'), ('age', int), ('weight', float)]
data1 = np.array([('Alice', 25, 55.5), ('Bob', 20, 68.0), ('Charlie', 30, 72.3)], dtype=dtype1)
print("1. Sorted by Age:\n", np.sort(data1, order='age'))

# Task 2:
dtype2 = [('x', int), ('y', int)]
pts = np.array([(1, 2), (4, 6)], dtype=dtype2)
dist = np.sqrt((pts['x'][0] - pts['x'][1])**2 + (pts['y'][0] - pts['y'][1])**2)
print("\n2. Euclidean distance:", dist)
```
