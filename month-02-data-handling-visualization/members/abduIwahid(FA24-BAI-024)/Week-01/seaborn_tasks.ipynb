# Seaborn Tasks

```python
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset("penguins")
titanic = sns.load_dataset("titanic")
tips = sns.load_dataset("tips")
```

## Task 1: Countplot

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=titanic, x="class", hue="sex")
plt.title("Titanic Passengers by Class and Sex")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
```

## Task 2: Histogram with KDE

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=penguins, x="flipper_length_mm", hue="species", kde=True)
plt.title("Flipper Length by Species")
plt.xlabel("Flipper Length (mm)")
plt.ylabel("Count")
plt.show()
```

## Task 3: Box vs Violin

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.boxplot(data=penguins, x="species", y="body_mass_g", hue="species", ax=axes[0])
axes[0].set_title("Box Plot of Body Mass by Species")
axes[0].set_xlabel("Species")
axes[0].set_ylabel("Body Mass (g)")
sns.violinplot(data=penguins, x="species", y="body_mass_g", hue="species", ax=axes[1])
axes[1].set_title("Violin Plot of Body Mass by Species")
axes[1].set_xlabel("Species")
axes[1].set_ylabel("Body Mass (g)")
plt.tight_layout()
plt.show()
```

## Task 4: Scatterplot

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=penguins, x="bill_length_mm", y="bill_depth_mm", hue="species", style="sex")
plt.title("Bill Length vs Bill Depth")
plt.xlabel("Bill Length (mm)")
plt.ylabel("Bill Depth (mm)")
plt.show()
```

## Task 5: Heatmap

```python
corr = penguins.select_dtypes(include="number").corr()

plt.figure(figsize=(7, 5))
sns.heatmap(corr, annot=True, cmap="coolwarm", center=0)
plt.title("Correlation Matrix of Penguin Measurements")
plt.show()
```

## Task 6: Pairplot

```python
g = sns.pairplot(penguins, hue="species")
g.figure.suptitle("Penguin Pairplot by Species", y=1.02)
plt.show()
```

## Task 7: lmplot

```python
sns.set_theme(style="whitegrid", palette="Set2", font_scale=1.1)

g = sns.lmplot(data=tips, x="total_bill", y="tip", col="time", row="smoker", height=3.5)
g.set_axis_labels("Total Bill", "Tip")
g.figure.suptitle("Total Bill vs Tip by Time and Smoker", y=1.03)
plt.show()
```
