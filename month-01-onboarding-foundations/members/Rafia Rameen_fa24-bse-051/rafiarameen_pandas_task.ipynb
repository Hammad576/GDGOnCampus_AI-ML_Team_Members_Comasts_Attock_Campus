**Assignment 01: Dataframe creation and indexing**

```python
import pandas as pd
import numpy as np

# Task 1:
df1 = pd.DataFrame(np.random.randint(1, 20, (6, 4)), columns=['A', 'B', 'C', 'D'])
df1 = df1.set_index('A')
print("1. Set index:\n", df1)

# Task 2:
df2 = pd.DataFrame(np.random.randint(1, 20, (3, 3)), index=['X', 'Y', 'Z'], columns=['A', 'B', 'C'])
print("2. Element at Y, B:", df2.loc['Y', 'B'])
```

**Assignment 02: Dataframe Operations**

```python
# Task 1:
df3 = pd.DataFrame(np.random.randint(1, 10, (5, 3)), columns=['Col1', 'Col2', 'Col3'])
df3['Product'] = df3['Col1'] * df3['Col2']
print("1. Product column:\n", df3)

# Task 2:
df4 = pd.DataFrame(np.random.randint(1, 10, (4, 3)), columns=['A', 'B', 'C'])
print("2. Row sum:\n", df4.sum(axis=1))
print("Column sum:\n", df4.sum(axis=0))
```

**Assignment 03: Data cleaning**

```python
# Task 1:
df5 = pd.DataFrame([[1, np.nan, 3], [4, 5, np.nan], [7, 8, 9]], columns=['A', 'B', 'C'])
print("1. Filled NaNs:\n", df5.fillna(df5.mean()))

# Task 2:
df6 = pd.DataFrame([[1, np.nan, 3], [4, 5, 6], [np.nan, 8, 9]], columns=['A', 'B', 'C'])
print("2. Dropped NaNs:\n", df6.dropna())
```

**Assignment 04: Data Aggregation**

```python
# Task 1:
df7 = pd.DataFrame({'Category': ['A', 'B', 'A', 'B', 'C'], 'Value': [10, 20, 30, 40, 50]})
print("1. Category Sum & Mean:\n", df7.groupby('Category')['Value'].agg(['sum', 'mean']))

# Task 2:
df8 = pd.DataFrame({'Product': ['P1', 'P2', 'P3'], 'Category': ['A', 'B', 'A'], 'Sales': [100, 200, 150]})
print("2. Total Sales:\n", df8.groupby('Category')['Sales'].sum())
```

**Assignment 05: Merging dataframes**

```python
# Task 1:
df_a = pd.DataFrame({'id': [1, 2], 'name': ['Alice', 'Bob']})
df_b = pd.DataFrame({'id': [1, 2], 'score': [85, 90]})
print("1. Merged DF:\n", pd.merge(df_a, df_b, on='id'))

# Task 2:
df_c = pd.DataFrame({'A': [1, 2]})
df_d = pd.DataFrame({'A': [3, 4]})
print("2. Concat Rows:\n", pd.concat([df_c, df_d], axis=0))
print("Concat Cols:\n", pd.concat([df_c, df_d], axis=1))
```

**Assignment 06: Time series analysis**

```python
# Task 1:
dates = pd.date_range('2021-01-01', periods=60)
df9 = pd.DataFrame({'Val': np.random.randint(1, 10, 60)}, index=dates)
print("1. Monthly Mean:\n", df9.resample('M').mean())

# Task 2:
dates2 = pd.date_range('2021-01-01', '2021-01-15')
df10 = pd.DataFrame({'Val': np.random.randint(1, 10, len(dates2))}, index=dates2)
print("2. Rolling Mean:\n", df10.rolling(7).mean())
```

**Assignment 07: Multi Index Dataframe**

```python
# Task 1:
index = pd.MultiIndex.from_tuples([('A', 1), ('A', 2), ('B', 1)])
df11 = pd.DataFrame({'Val': [10, 20, 30]}, index=index)
print("1. Sliced 'A':\n", df11.loc['A'])

# Task 2:
idx = pd.MultiIndex.from_tuples([('Cat1', 'Sub1'), ('Cat1', 'Sub2'), ('Cat2', 'Sub1')])
df12 = pd.DataFrame({'Data': [5, 10, 15]}, index=idx)
print("2. MultiIndex Sum:\n", df12.groupby(level=0).sum())
```

**Assignment 08: Pivot Tables**

```python
# Task 1:
df13 = pd.DataFrame({'Date': ['Day1', 'Day1', 'Day2'], 'Category': ['A', 'B', 'A'], 'Value': [10, 20, 30]})
print("1. Pivot 1:\n", df13.pivot_table(index='Date', columns='Category', values='Value', aggfunc='sum'))

# Task 2:
df14 = pd.DataFrame({'Year': [2020, 2020, 2021], 'Quarter': ['Q1', 'Q2', 'Q1'], 'Revenue': [100, 200, 150]})
print("2. Pivot 2:\n", df14.pivot_table(index='Year', columns='Quarter', values='Revenue', aggfunc='mean'))
```

**Assignment 09: Applying Functions**

```python
# Task 1:
df15 = pd.DataFrame([[1, 2], [3, 4]], columns=['A', 'B'])
print("1. Doubled:\n", df15.apply(lambda x: x * 2))

# Task 2:
df16 = pd.DataFrame([[1, 2, 3], [4, 5, 6]], columns=['A', 'B', 'C'])
df16['Sum'] = df16.apply(lambda row: row.sum(), axis=1)
print("2. Row sum added:\n", df16)
```
