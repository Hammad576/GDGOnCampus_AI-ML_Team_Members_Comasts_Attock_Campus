## Assignment 4: Data Aggregation

1. Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.

```python
df = pd.DataFrame({"Category": rng.choice(["A", "B", "C"], 12), "Value": rng.integers(1, 100, 12)})
display(df)
display(df.groupby("Category")["Value"].agg(["sum", "mean"]))
```

2. Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.

```python
df = pd.DataFrame({"Product": rng.choice(["Pen", "Book", "Bag"], 10),
                   "Category": rng.choice(["A", "B", "C"], 10),
                   "Sales": rng.integers(100, 1000, 10)})
display(df)
display(df.groupby("Category")["Sales"].sum())
```

## Assignment 6: Time Series Analysis

1. Create a Pandas DataFrame with a datetime index and one column filled with random integers. Resample the DataFrame to compute the monthly mean of the values.

```python
dates = pd.date_range("2021-01-01", "2021-12-31", freq="D")
df = pd.DataFrame({"Value": rng.integers(1, 100, len(dates))}, index=dates)
# MS labels each monthly group by its first day.
display(df.resample("MS").mean())
```

2. Create a Pandas DataFrame with a datetime index ranging from '2021-01-01' to '2021-12-31' and one column filled with random integers. Compute the rolling mean with a window of 7 days.

```python
dates = pd.date_range("2021-01-01", "2021-12-31", freq="D")
df = pd.DataFrame({"Value": rng.integers(1, 100, len(dates))}, index=dates)
# Daily data: seven observations form a seven-day window.
df["RollingMean"] = df["Value"].rolling(window=7).mean()
display(df.head(10))  # The first six rows lack a full window.
```

## Assignment 7: MultiIndex DataFrame

1. Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform some basic indexing and slicing operations on the MultiIndex DataFrame.

```python
index = pd.MultiIndex.from_product([["A", "B"], [1, 2, 3]], names=["Group", "Item"])
df = pd.DataFrame({"Value": rng.integers(1, 100, 6)}, index=index).sort_index()
display(df)
print("Group A:")
display(df.loc["A"])
print("Single value:", df.loc[("A", 2), "Value"])
print("Items 1 to 2 in both groups:")
display(df.loc[pd.IndexSlice[:, 1:2], :])
```

2. Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.

```python
index = pd.MultiIndex.from_tuples(
    [("A", "X"), ("A", "X"), ("A", "Y"), ("B", "X"), ("B", "Y"), ("B", "Y")],
    names=["Category", "SubCategory"])
df = pd.DataFrame({"Value": rng.integers(1, 100, 6)}, index=index)
display(df)
display(df.groupby(level=["Category", "SubCategory"]).sum())
```

## Assignment 8: Pivot Tables

1. Create a Pandas DataFrame with columns 'Date', 'Category', and 'Value'. Create a pivot table to compute the sum of 'Value' for each 'Category' by 'Date'.

```python
df = pd.DataFrame({"Date": pd.to_datetime(["2021-01-01"] * 3 + ["2021-01-02"] * 3),
                   "Category": ["A", "A", "B", "A", "B", "B"],
                   "Value": rng.integers(1, 100, 6)})
display(df)
display(df.pivot_table(index="Date", columns="Category", values="Value", aggfunc="sum", fill_value=0))
```

2. Create a Pandas DataFrame with columns 'Year', 'Quarter', and 'Revenue'. Create a pivot table to compute the mean 'Revenue' for each 'Quarter' by 'Year'.

```python
df = pd.DataFrame({"Year": np.repeat([2021, 2022], 8),
                   "Quarter": np.tile(np.repeat([1, 2, 3, 4], 2), 2),
                   "Revenue": rng.integers(1000, 5000, 16)})
display(df)
display(df.pivot_table(index="Year", columns="Quarter", values="Revenue", aggfunc="mean"))
```

## Assignment 9: Applying Functions
