## Assignment 3

```python
df = pd.DataFrame(np.random.randint(1, 50, size=(5, 3)), columns=["A", "B", "C"])
df = df.astype(float)
df.loc[1, "A"] = np.nan
df.loc[3, "B"] = np.nan
df.loc[0, "C"] = np.nan
print(df)
df = df.fillna(df.mean())
print(df)
```

```python
df = pd.DataFrame(np.random.randint(1, 50, size=(6, 4)), columns=["A", "B", "C", "D"])
df = df.astype(float)
df.loc[0, "B"] = np.nan
df.loc[2, "D"] = np.nan
df.loc[4, "A"] = np.nan
print(df)
df = df.dropna()
print(df)
```

## Assignment 4

```python
df = pd.DataFrame({
    "Category": np.random.choice(["A", "B", "C"], size=10),
    "Value": np.random.randint(1, 100, size=10)
})
print(df)
print(df.groupby("Category")["Value"].agg(["sum", "mean"]))
```

```python
df = pd.DataFrame({
    "Product": ["Pen", "Book", "Shirt", "Pencil", "Jeans", "Notebook"],
    "Category": ["Stationery", "Stationery", "Clothing", "Stationery", "Clothing", "Stationery"],
    "Sales": np.random.randint(100, 1000, size=6)
})
print(df)
print(df.groupby("Category")["Sales"].sum())
```

## Assignment 5

```python
df1 = pd.DataFrame({"ID": [1, 2, 3, 4], "Name": ["Ali", "Sara", "Ahmed", "Zara"]})
df2 = pd.DataFrame({"ID": [1, 2, 3, 4], "Marks": [85, 90, 78, 88]})
merged = pd.merge(df1, df2, on="ID")
print(merged)
```

```python
df1 = pd.DataFrame({"A": [1, 2, 3], "B": [4, 5, 6]})
df2 = pd.DataFrame({"C": [7, 8, 9], "D": [10, 11, 12]})
print(pd.concat([df1, df2], axis=0))
print(pd.concat([df1, df2], axis=1))
```

## Assignment 6

```python
dates = pd.date_range(start="2021-01-01", periods=100, freq="D")
df = pd.DataFrame({"Value": np.random.randint(1, 100, size=100)}, index=dates)
print(df.head())
print(df.resample("ME").mean())
```

```python
dates = pd.date_range(start="2021-01-01", end="2021-12-31", freq="D")
df = pd.DataFrame({"Value": np.random.randint(1, 100, size=len(dates))}, index=dates)
df["Rolling Mean"] = df["Value"].rolling(window=7).mean()
print(df.head(10))
```

## Assignment 7

```python
index = pd.MultiIndex.from_tuples([
    ("Class A", "Ali"),
    ("Class A", "Sara"),
    ("Class B", "Ahmed"),
    ("Class B", "Zara")
], names=["Class", "Student"])
df = pd.DataFrame({"Marks": [80, 90, 75, 85]}, index=index)
print(df)
print(df.loc["Class A"])
print(df.loc[("Class B", "Zara")])
print(df.loc["Class A":"Class B"])
```

```python
index = pd.MultiIndex.from_tuples([
    ("Electronics", "Mobile"),
    ("Electronics", "Laptop"),
    ("Electronics", "Mobile"),
    ("Clothing", "Shirt"),
    ("Clothing", "Jeans"),
    ("Clothing", "Shirt")
], names=["Category", "SubCategory"])
df = pd.DataFrame({"Value": np.random.randint(1, 100, size=6)}, index=index)
print(df)
print(df.groupby(["Category", "SubCategory"]).sum())
```

## Assignment 8

```python
df = pd.DataFrame({
    "Date": ["2021-01-01", "2021-01-01", "2021-01-02", "2021-01-02", "2021-01-03", "2021-01-03"],
    "Category": ["A", "B", "A", "B", "A", "B"],
    "Value": np.random.randint(1, 100, size=6)
})
print(df)
pivot = pd.pivot_table(df, values="Value", index="Date", columns="Category", aggfunc="sum")
print(pivot)
```

```python
df = pd.DataFrame({
    "Year": [2020, 2020, 2020, 2020, 2021, 2021, 2021, 2021],
    "Quarter": ["Q1", "Q2", "Q3", "Q4", "Q1", "Q2", "Q3", "Q4"],
    "Revenue": np.random.randint(1000, 5000, size=8)
})
print(df)
pivot = pd.pivot_table(df, values="Revenue", index="Year", columns="Quarter", aggfunc="mean")
print(pivot)
```

```python
df = pd.DataFrame(np.random.randint(1, 20, size=(5, 3)), columns=["A", "B", "C"])
print(df)

def double(x):
    return x * 2

print(df.apply(double))
```
