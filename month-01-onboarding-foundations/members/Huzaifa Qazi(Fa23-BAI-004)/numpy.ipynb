```python
dt = np.dtype([('name', 'U10'), ('age', 'i4'), ('weight', 'f4')])
data = np.array([('Ali', 25, 70.5),
                 ('Sara', 22, 55.2),
                 ('Ahmed', 30, 80.1),
                 ('Fatima', 19, 60.0)], dtype=dt)
print(np.sort(data, order='age'))
```

```python
dt = np.dtype([('x', int), ('y', int)])
pts = np.array([(0, 0), (3, 4), (6, 8), (1, 1)], dtype=dt)
n = len(pts)
for i in range(n):
    for j in range(i + 1, n):
        dx = pts['x'][i] - pts['x'][j]
        dy = pts['y'][i] - pts['y'][j]
        d = np.sqrt(dx**2 + dy**2)
        print("point", i, "and point", j, "->", d)
```

```python
a = np.random.randint(1, 21, (4, 4))
print(a)
m = np.ma.masked_greater(a, 10)
print(m)
print("sum:", m.sum())
```

```python
a = np.random.randint(1, 21, (3, 3)).astype(float)
mask = np.zeros((3, 3), dtype=bool)
for i in range(3):
    mask[i, i] = True   # mask the diagonal
m = np.ma.masked_array(a, mask=mask)
avg = m.mean()
print(m)
print("mean of unmasked:", avg)
print(m.filled(avg))
```
